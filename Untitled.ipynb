```python
import akshare as ak
import matplotlib.pyplot as plt
import mplfinance as mpf

# 定义获取股票日线数据的函数
def get_stock_daily_data(stock_code, start_date, end_date):
    # 使用akshare的stock_zh_a_daily函数获取指定股票在指定日期范围内的日线数据
    stock_data = ak.stock_zh_a_daily(symbol=stock_code, start_date=start_date, end_date=end_date, adjust="qfq")
    # 将日期列转换为datetime类型，并设置为索引
    stock_data.index = stock_data['date'].astype('datetime64[ns]')
    # 选取绘图所需的列：开盘价、最高价、最低价、收盘价、成交量
    columns = ['open', 'high', 'low', 'close', 'volume']
    return stock_data[columns]

# 定义绘制K线图的函数
def plot_stock_k_line(stock_data):
    # 设置图片清晰度
    plt.rcParams['figure.dpi'] = 300
    # 使用mplfinance的plot函数绘制K线图
    mpf.plot(stock_data, type='candle', mav=(5, 10, 20), volume=True)
    plt.show()

if __name__ == "__main__":
    # 股票代码，这里以贵州茅台（600519）为例
    stock_code = "sh600519"
    # 开始日期
    start_date = "20240801"
    # 结束日期
    end_date = "20250122"
    # 调用函数获取股票日线数据
    stock_daily_data = get_stock_daily_data(stock_code, start_date, end_date)
    # 调用函数绘制K线图
    plot_stock_k_line(stock_daily_data)
```
